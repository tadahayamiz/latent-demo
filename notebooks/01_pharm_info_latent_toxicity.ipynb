```python
#@title 5. 因子数の比較 { display-mode: "form" }
max_factors = 5  #@param {type:"slider", min:2, max:8, step:1}
summary = compare_factor_numbers(
    scenario=scenario,
    model=model,
    max_factors=max_factors,
    standardize=standardize,
)
```

## 5. まとめ

- 観測できるのは検査値であり、本当に知りたいのは背後の状態である。

- 因子分析では、共通因子と変数ごとの固有ノイズを分けて考える。
